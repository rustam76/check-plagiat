```python
# ===============================================
# 🧬 GENERATOR DATASET PLAIGIAT OTOMATIS
# ===============================================
import pandas as pd
import random

# 1️⃣ Load dataset sumber
df_source = pd.read_csv("dataset_pairs_clean.csv")

# Pastikan hanya ambil kolom utama
df_source = df_source[['title_1', 'abstract_1']].drop_duplicates().reset_index(drop=True)
print("Jumlah teks unik sumber:", len(df_source))

# ===============================================
# 2️⃣ Buat pasangan positif (label=1)
# ===============================================
# Positif = judul dan abstrak dari baris yang sama
positive_pairs = []
for _, row in df_source.iterrows():
    positive_pairs.append({
        'title_1': row['title_1'],
        'abstract_1': row['abstract_1'],
        'title_2': row['title_1'],
        'abstract_2': row['abstract_1'],
        'label': 1
    })

print("✅ Jumlah pasangan positif (plagiat):", len(positive_pairs))

# ===============================================
# 3️⃣ Buat pasangan negatif (label=0)
# ===============================================
negative_pairs = []
titles = df_source['title_1'].tolist()
abstracts = df_source['abstract_1'].tolist()

for _ in range(len(positive_pairs)):
    t1, t2 = random.sample(titles, 2)
    a1, a2 = random.sample(abstracts, 2)
    negative_pairs.append({
        'title_1': t1,
        'abstract_1': a1,
        'title_2': t2,
        'abstract_2': a2,
        'label': 0
    })

print("✅ Jumlah pasangan negatif (non-plagiat):", len(negative_pairs))

# ===============================================
# 4️⃣ Gabungkan dan acak ulang
# ===============================================
df_generated = pd.DataFrame(positive_pairs + negative_pairs)
df_generated = df_generated.sample(frac=1, random_state=42).reset_index(drop=True)

# ===============================================
# 5️⃣ Simpan hasil
# ===============================================
df_generated.to_csv("dataset_pairs_generated.csv", index=False)
print("\n🎯 Dataset baru berhasil dibuat dan disimpan ke dataset_pairs_generated.csv")
print("Jumlah total data:", len(df_generated))
print(df_generated.head(3))
```
